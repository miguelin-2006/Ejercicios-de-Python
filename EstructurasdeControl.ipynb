```python
#Ciclo while
x = 1
while x < 5:
    print(x)
    x += 1
```

```python
#iterando con while
nombre = "Pablo"
k = 0
while k < len(nombre):
    print(nombre[k])
    k += 1
```

```python
#combina el uso de while con el condicional if-else
from random import randint

print("¡Bienvenido a Adivina el Número!")
n = randint(1,10) # Genera un entero aleatorio en el intervalo [1,10]
k = 1 # número de intentos

while True:
    x = int( input("Ingrese un entero entre 1 y 10: ") )
    if x == n:
        print(f"Has adivinado después de {k} intentos")
        break
    else:
        print(f"{x} no es el número, intenta nuevamente\n")
    k += 1
```

```python
#Enumerate 
lista = ["A", "B", "C"]

for l in lista:
    print(l)

print('Sacamos el indice')
indice = 0
for l in lista:
    print(indice, l)
    indice += 1
print('Con enumerate sacamos el indice')

indice = 0
for indice, l in enumerate(lista):
    print(indice, l)

```

```python
#Enumerate:  no se limita únicamente a bucles for. Podemos convertir el tipo enumerate en una lista de tuplas,
#donde cada una contiene un elemento de la colección inicial y el índice asociado.
lista = ["A", "B", "C"]

en = list(enumerate(lista))
print(en)
```

```python
#match
mes = 4
match mes:
    case 12 | 1 | 2: print("Invierno")
    case 3 | 4 | 5: print("Primavera")
    case 6 | 7 | 8: print("Verano")
    case 9 | 10 | 11: print("Otoño")
    case _: print("Error")
# Primavera

coordenada = (0, 1)
match coordenada:
    case (0, 0):
        print("Coordenada en origen")
    case (x, 0):
        print(f"Coordenada en eje x: {x}")
    case (0, y):
        print(f"Coordenada en eje y: {y}")
    case (x, y):
        print(f"Coordenada en: {x}, {y}")
    case _:
        print("Error")

```

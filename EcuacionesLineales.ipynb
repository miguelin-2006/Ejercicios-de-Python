```python
#Sistema de ecuaciones lineales en Python
'''
El sistema que vamos a resolver es:
x + 2y -z +3t = -8
2x + 2z -t = 13
-x +y +z -t = 8
3x + 3y -z +2t = -1
'''
from numpy  import array, linalg

#En la matriz colocamos los coeficientes de las incógnitas
matriz = array([
    [1 , 2,-1, 3],
    [2 , 0, 2,-1],
    [-1, 1, 1,-1],
    [3 , 3,-1, 2]
    ])

#En el vector colocamos los términos independientes
vector=array([-8,13,8,-1])

#Obtenemos la solución con lineal
solucion = linalg.solve(matriz, vector)

print('La solución es: ')
for i in range(0, len(solucion)):
    print(f'x{i}:', solucion[i])


```
